# Bellwether — skilled strategists on EXTERNAL-FACT markets

Pivot away from the latency/updown pool. These markets resolve against a
**knowable external fact** — elections, policy, sports series, crypto-event
resolution, geopolitics. Edge here is **information / modeling / discipline**,
not speed. Updown/intraday families are excluded up front.

Funnel, per (account, category): **eligibility** (>=10 distinct resolved
markets) → **persistence** (net-positive in most sub-periods) → **type** by
entry timing (predictor / momentum-rider / favorite-farmer) → **type-appropriate
skill test** → **rank** survivors by P&L and % return. A market-calibrated null
control must produce no ranked survivors.

## Step 0 — recon: studyable population per category
Thin categories (e.g. Spotify/TV) are flagged 'no studyable population' rather
than forced. Needs a loaded pool (see `docs/EXTERNAL_FACT_STRATEGISTS.md`).

```python
from bellwether_analytics.core import load_trades_df
from bellwether_analytics.skill import category_recon

trades = load_trades_df(platform='polymarket')
category_recon(trades)
```

## The funnel + ranking
Every eligible account with its distinct-resolved-market count, sub-period
persistence, entry-timing type, type-appropriate skill result, P&L, % return,
survival verdict, and the information-vs-modeling hook.

```python
from bellwether_analytics.skill import rank_strategists, rank_within_type

funnel = rank_strategists(trades)
display(funnel)
print('survivors, ranked within type by P&L and % return:')
rank_within_type(funnel)
```
